# 2. Calculate successful and failed login counts

```python
print("Successful Login:", (df["Login Status"] == "SUCCESS").sum())

print("Failed Login:", (df["Login Status"] == "FAILED").sum())
```

# Compute failed login rates by account type and authentication method

```python

total = df.groupby("Authentication Method")["Login Status"].count()

failed = df[df["Login Status"] == "FAILED"].groupby("Authentication Method")["Login Status"].count()

failed = failed.reindex(total.index, fill_value=0)

rate = (failed / total) * 100

print("Failed Login Rate by Authentication Method")
print(rate)
```

# Analyze login activity by hour of the day and day of the week

```python
df["Hour"] = df["Timestamp"].dt.hour

hour_login = df["Hour"].value_counts().sort_index()

print("Login Activity by Hour")
print(hour_login)

df["Day"] = df["Timestamp"].dt.day_name()

days = ["Monday", "Tuesday", "Wednesday", "Thursday",
        "Friday", "Saturday", "Sunday"]

day_login = df["Day"].value_counts().reindex(days, fill_value=0)

print("\nLogin Activity by Day")
print(day_login)

print("\nBusiest Hour:", hour_login.idxmax(), "O'clock")
print("Busiest Day:", day_login.idxmax())
```

#  Identify unusual login patterns, such as repeated failures or unexpected access times. 

```python
# 1. Repeated Failed Login Attempts
failed_login = df[df["Login Status"] == "FAILED"]

failed_count = failed_login.groupby("User ID").size()

print("Repeated Failed Login Attempts:")
print(failed_count[failed_count >= 3])


# 2. Unexpected Login Times (12 AM to 5 AM)
unexpected_login = df[
    (df["Timestamp"].dt.hour >= 0) &
    (df["Timestamp"].dt.hour < 5)
]

print("\nUnexpected Login Activities:")
print(unexpected_login[["User ID", "Timestamp", "Login Status"]])


# 3. Total Unusual Login Attempts
print("\nTotal Repeated Failed Login Users:",
      (failed_count >= 3).sum())

print("Total Unexpected Login Attempts:", len(unexpected_login))
```

# Create a heatmap showing login activity by day and hour

```python

df["Hour"] = df["Timestamp"].dt.hour
df["Day"] = df["Timestamp"].dt.day_name()

days = ["Monday", "Tuesday", "Wednesday", "Thursday",
        "Friday", "Saturday", "Sunday"]

heatmap_data = df.pivot_table(
    index="Day",
    columns="Hour",
    values="User ID",
    aggfunc="count",
    fill_value=0
)

heatmap_data = heatmap_data.reindex(days, fill_value=0)

plt.figure(figsize=(15, 6))

plt.imshow(heatmap_data, cmap="YlOrRd", aspect="auto")


plt.xticks(range(len(heatmap_data.columns)), heatmap_data.columns)
plt.yticks(range(len(heatmap_data.index)), heatmap_data.index)

plt.colorbar(label="Number of Logins")

plt.title("Login Activity by Day and Hour")
plt.xlabel("Hour of the Day")
plt.ylabel("Day of the Week")

plt.show()
```

# Create a stacked bar chart comparing successful and failed logins.

```python
import matplotlib.pyplot as plt

success = (df["Login Status"] == "SUCCESS").sum()
failed = (df["Login Status"] == "FAILED").sum()

# Create Stacked Bar Chart
plt.bar("Login Activity", success,  label="Successful")
plt.bar("Login Activity", failed, bottom=success, label="Failed")

# Add Labels
plt.title("Successful vs Failed Login Attempts")
plt.xlabel("Login Status")
plt.ylabel("Number of Logins")
plt.legend()

plt.show()
```

# Create a line chart showing failed login attempts over time.

```python

failed_login = df[df["Login Status"] == "FAILED"].copy()

# Step 3: Group by Date and Count Failed Logins
failed_over_time = failed_login.groupby(
    failed_login["Timestamp"].dt.date
).size()

# Step 4: Create Line Chart
plt.figure(figsize=(12, 5))

plt.plot(failed_over_time.index, failed_over_time.values,
         marker="o", color="red")

# Step 5: Add Labels
plt.title("Failed Login Attempts Over Time")
plt.xlabel("Date")
plt.ylabel("Number of Failed Logins")
plt.xticks(rotation=45)
plt.grid()

plt.tight_layout()
plt.show()
```
